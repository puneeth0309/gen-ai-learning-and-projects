```python
import opendatasets as od
import pandas as pd
od.download("https://www.kaggle.com/datasets/lakshmi25npathi/imdb-dataset-of-50k-movie-reviews")
df=pd.read_csv("imdb-dataset-of-50k-movie-reviews/IMDB Dataset.csv")
print(df.head())
print(df.shape
      )
```

```python
df['review']=df['review'].str.lower()
print(df['review'][3])
```

```python
import re
def remove_html_tags(text):
    clean = re.compile('<.*?>')
    return re.sub(clean, '', text)
df['review']=df['review'].apply(remove_html_tags) 
print(df['review'][3])
```

```python
from textblob import TextBlob
import pandas as pd

# sample test
text = "I ame wery vappay"
txt = TextBlob(text)
print(txt.correct())


# function
def correct_spelling(text):
    if not isinstance(text, str):
        return text
    return str(TextBlob(text).correct())


# -------- use only 100 rows --------
df_small = df.head(100).copy()

df_small['review'] = df_small['review'].apply(correct_spelling)

print(df_small['review'][3])
```

```python
import pandas as pd
from nltk.corpus import stopwords
import nltk

# Download stopwords (runs once, then skips automatically)
nltk.download('stopwords')

# Sample DataFrame
df = pd.DataFrame({
    'review': [
        'This is a good movie',
        'This is a bad movie',
        'This is an excellent movie',
        'This is a terrible movie'
    ]
})

# Load stopwords
stop_words = set(stopwords.words('english'))

# Function to remove stopwords
def remove_stopwords(text):
    return " ".join([word for word in text.split() if word.lower() not in stop_words])

# Apply function
df['review'] = df['review'].apply(remove_stopwords)

# Print entire dataframe
print("Full DataFrame:\n", df)

# Print 3rd row (index 3)
print("\n3rd row:")
print(df.iloc[3])          # safest way

# OR only review column of 3rd row
print("\n3rd row review:")
print(df['review'].iloc[3])
```
