```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
df=pd.DataFrame({'review':['This is a good movie','This is a bad movie','This is an excellent movie','This is a terrible movie'],'rating':[1,1,0,0]})
cv=CountVectorizer()
x=cv.fit_transform(df['review'])
print(cv.vocabulary_)
print(x.toarray())
```

```python
cv=CountVectorizer(ngram_range=(2,2))
x=cv.fit_transform(df['review'])
print(cv.vocabulary_)
print(x.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer(ngram_range=(2,2))
x=tfidf.fit_transform(df['review'])
print(tfidf.vocabulary_)
print(x.toarray())
```

```python
import os
import nltk
from nltk.tokenize import sent_tokenize
from gensim.utils import simple_preprocess
from gensim.models import Word2Vec

# Download tokenizer resources (run once)
nltk.download('punkt')
nltk.download('punkt_tab')

# Dataset folder path
data_path = r'game-of-thrones-book-dataset'

story = []

# Read all txt files
for filename in os.listdir(data_path):
    if filename.endswith('.txt'):
        file_path = os.path.join(data_path, filename)

        with open(file_path, 'r', encoding='latin-1') as f:
            corpus = f.read()

            # Split into sentences
            sentences = sent_tokenize(corpus)

            # Convert each sentence into cleaned word tokens
            for sent in sentences:
                tokens = simple_preprocess(sent)
                story.append(tokens)

print(story[:5])
```

```python
model = Word2Vec(
    sentences=story,
    vector_size=100,
    window=5,
    min_count=2,
    workers=4
)

# Example: similar words
print(model.wv.most_similar('viserys')) 
```

```python
print(model.wv['deep'].shape)
from sklearn.decomposition import PCA
pca=PCA(n_components=3)
x=pca.fit_transform(model.wv.get_normed_vectors())
print(x)
index = model.wv.key_to_index['deep']
print(x[index])
print(x.shape)
```
